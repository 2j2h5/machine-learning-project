```python
import sys
sys.path.insert(0, '../src')
import matplotlib.pyplot as plt
import lab01

raw = lab01.load_data('../data/cafe_sales.csv')
clean = lab01.clean_data(raw)
raw.head()
```

## Plot 1 — Missing values per column (raw data)

```python
# ============ TODO (Plot 1) ============
from pathlib import Path

figure_dir = Path('../../submissions/lab01/figures')
figure_dir.mkdir(parents=True, exist_ok=True)
plt.rcParams.update({'font.size': 10, 'axes.spines.top': False, 'axes.spines.right': False})
missing = lab01.summarize_missing(raw)
fig, ax = plt.subplots(figsize=(7.2, 2.7), layout='constrained')
bars = ax.bar(missing.index, missing.values, color='#277C8E', width=0.55)
ax.bar_label(bars, padding=4)
ax.set(title='Missing values before cleaning (412 rows)', xlabel='Column', ylabel='Missing entries')
ax.set_ylim(0, missing.max() * 1.25)
ax.set_axisbelow(True)
ax.grid(axis='y', alpha=0.2)
fig.savefig(figure_dir / 'figure1.png', dpi=180)
plt.show()
# =======================================
```

## Plot 2 — Distribution of `total_price` (clean data)

```python
# ============ TODO (Plot 2) ============
from matplotlib.ticker import StrMethodFormatter

# The long right tail includes plausible bulk orders; mean exceeds median.
fig, ax = plt.subplots(figsize=(7.2, 3.1), layout='constrained')
ax.hist(clean['total_price'], bins=30, color='#277C8E', edgecolor='white')
ax.set(title='Order totals after cleaning (400 rows)', xlabel='Total price (dataset currency units)', ylabel='Orders')
ax.xaxis.set_major_formatter(StrMethodFormatter('{x:,.0f}'))
ax.tick_params(axis='x', labelsize=8)
ax.set_axisbelow(True)
ax.grid(axis='y', alpha=0.2)
zoom = ax.inset_axes([0.37, 0.38, 0.59, 0.49])
zoom.hist(clean['total_price'], bins=30, range=(0, 35000), color='#4B9FAD', edgecolor='white')
zoom.set_title('Zoom: totals from 0 to 35,000', fontsize=9)
zoom.set_ylabel('Orders', fontsize=8)
zoom.tick_params(labelsize=8)
zoom.xaxis.set_major_formatter(StrMethodFormatter('{x:,.0f}'))
fig.savefig(figure_dir / 'figure2.png', dpi=180)
plt.show()
# =======================================
```

## Plot 3 — `customer_rating` by category (clean data)

```python
# ============ TODO (Plot 3) ============
fig, ax = plt.subplots(figsize=(7.2, 3.1), layout='constrained')
clean.boxplot(column='customer_rating', by='category', ax=ax, grid=False, patch_artist=True, boxprops={'facecolor': '#C7E2E5'}, medianprops={'color': '#AD542F', 'linewidth': 1.6})
fig.suptitle('')
ax.set(title='Customer ratings by category after cleaning', xlabel='Category (cleaned order count)', ylabel='Customer rating')
counts = clean.groupby('category').size()
ax.set_xticklabels([f'{category}\n(n={count})' for category, count in counts.items()])
ax.set_ylim(0.8, 5.2)
ax.set_axisbelow(True)
ax.grid(axis='y', alpha=0.2)
fig.savefig(figure_dir / 'figure3.png', dpi=180)
plt.show()
# =======================================
```
