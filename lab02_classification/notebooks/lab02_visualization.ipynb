## Plot 1 — Class balance

```python
# ============ TODO (Plot 1) ============
# Bar chart of y.value_counts(). Title + axis labels.
# =======================================
from pathlib import Path
FIG_DIR = Path('../../submissions/lab02/figures')
FIG_DIR.mkdir(parents=True, exist_ok=True)

counts = y.value_counts().sort_index()
fig, ax = plt.subplots(figsize=(6, 3.6))
bars = ax.bar(['0 (showed up)', '1 (no-show)'], counts.values, color=['#277C8E', '#D9822B'])
for bar, n in zip(bars, counts.values):
    ax.text(bar.get_x() + bar.get_width() / 2, n, f'{n} ({n / len(y):.1%})',
            ha='center', va='bottom')
ax.set_title('Class balance of no_show')
ax.set_xlabel('no_show')
ax.set_ylabel('Number of appointments')
ax.set_ylim(0, counts.max() * 1.15)
fig.tight_layout()
fig.savefig(FIG_DIR / 'figure1.png', dpi=150)
plt.show()
```

## Plot 2 — One feature, split by class

```python
# ============ TODO (Plot 2) ============
# Overlaid histograms of X['days_ahead'] for y==0 vs y==1 (alpha=0.5, legend).
# Comment: does this feature separate the classes?
# =======================================
import numpy as np
bins = np.linspace(X['days_ahead'].min(), X['days_ahead'].max(), 25)
fig, ax = plt.subplots(figsize=(6, 3.6))
ax.hist(X.loc[y == 0, 'days_ahead'], bins=bins, alpha=0.5, color='#277C8E', label='0 (showed up)')
ax.hist(X.loc[y == 1, 'days_ahead'], bins=bins, alpha=0.5, color='#D9822B', label='1 (no-show)')
ax.set_title('days_ahead distribution by class')
ax.set_xlabel('days_ahead (days between booking and appointment)')
ax.set_ylabel('Number of appointments')
ax.legend()
fig.tight_layout()
fig.savefig(FIG_DIR / 'figure2.png', dpi=150)
plt.show()
# Comment: no-shows shift toward longer lead times, but the two histograms overlap heavily,
# so days_ahead alone cannot separate the classes.
print(X.groupby(y)['days_ahead'].describe())
```

## Plot 3 — Model comparison

```python
# ============ TODO (Plot 3) ============
# Grouped bar chart: accuracy and f1 for baseline / logreg / knn (use `results`).
# =======================================
models = ['baseline', 'logreg', 'knn']
metrics = ['accuracy', 'f1']
colors = {'accuracy': '#277C8E', 'f1': '#D9822B'}
x = np.arange(len(models))
width = 0.36
fig, ax = plt.subplots(figsize=(6, 3.6))
for i, metric in enumerate(metrics):
    values = [results[m][metric] for m in models]
    bars = ax.bar(x + (i - 0.5) * width, values, width, label=metric, color=colors[metric])
    for bar, v in zip(bars, values):
        ax.text(bar.get_x() + bar.get_width() / 2, v, f'{v:.3f}', ha='center', va='bottom', fontsize=8)
ax.set_xticks(x, ['baseline (majority)', 'logistic regression', 'kNN (k=15)'])
ax.set_title('Validation accuracy and F1 by model')
ax.set_xlabel('Model')
ax.set_ylabel('Score')
ax.set_ylim(0, 1.05)
ax.legend()
fig.tight_layout()
fig.savefig(FIG_DIR / 'figure3.png', dpi=150)
plt.show()
```
